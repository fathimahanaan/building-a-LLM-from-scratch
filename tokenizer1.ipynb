```python
import re  # Imports Python's regular expression module, used for splitting and modifying text


class SimpleTokenizerV1:
    # Creates a class called SimpleTokenizerV1.
    # A tokenizer converts text into numbers (token IDs) and numbers back into text.

    def __init__(self, vocab):
        # __init__ runs automatically when we create a tokenizer object.
        # vocab is a dictionary containing words/tokens and their corresponding IDs.

        self.str_to_int = vocab
        # Stores the vocabulary dictionary.
        # Example:
        # {"hello": 0, "world": 1, "!": 2}

        self.int_to_str = {i: s for s, i in vocab.items()}
        # Creates the reverse dictionary.
        # Instead of:
        # {"hello": 0}
        # We get:
        # {0: "hello"}
        #
        # This allows us to convert token IDs back into words.


    def encode(self, text):
        # encode() converts normal text into a list of token IDs.
        # Example:
        # "Hello world!"
        # becomes something like:
        # [5, 10, 2]

        preprocessed = re.split(
           
        # Splits the text using regular expressions.
        #
        # It separates punctuation such as:
        # , . : ; ? _ ! " ( )
        #
        # It also separates "--" and whitespace.
        #
        # For example:
        # "Hello, world!"
        # can become:
        # ["Hello", ",", "", " ", "world", "!", ""]


        preprocessed = [
            item.strip()
            for item in preprocessed
            if item.strip()
        ]
        # Removes empty strings and unnecessary whitespace.
        #
        # For example:
        # ["Hello", ",", "", " ", "world", "!"]
        #
        # becomes:
        # ["Hello", ",", "world", "!"]


        ids = [self.str_to_int[s] for s in preprocessed]
        # Converts every token into its corresponding number.
        #
        # Example vocabulary:
        # {"Hello": 5, ",": 2, "world": 10, "!": 3}
        #
        # ["Hello", ",", "world", "!"]
        #
        # becomes:
        # [5, 2, 10, 3]


        return ids
        # Returns the list of token IDs.


    def decode(self, ids):
        # decode() does the opposite of encode().
        # It converts token IDs back into text.
        #
        # Example:
        # [5, 2, 10, 3]
        # becomes:
        # "Hello , world !"

        text = " ".join([self.int_to_str[i] for i in ids])
        # Converts every number back into its corresponding token.
        #
        # Then joins all the tokens together with spaces.
        #
        # Example:
        # ["Hello", ",", "world", "!"]
        # becomes:
        # "Hello , world !"


        # Replace spaces before the specified punctuations
        text = re.sub(
            r'\s+([,.?!"()\'])',
            r'\1',
            text
        )
        # Removes spaces before punctuation.
        #
        # Before:
        # "Hello , world !"
        #
        # After:
        # "Hello, world!"
        #
        # \s+ means one or more spaces.
        # ([,.?!"()']) means one of these punctuation characters.
        # \1 puts the punctuation back without the space.


        return text
        # Returns the final decoded text.
```

```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
                                
        preprocessed = [
            item.strip() for item in preprocessed if item.strip()
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = { i:s for s,i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        preprocessed = [
            item if item in self.str_to_int 
            else "<|unk|>" for item in preprocessed
        ]

        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```
